```python
def calculate_probability(event_count, total_outcomes):
    return event_count / total_outcomes

probability_heads = calculate_probability(1, 2)

print("Probability of getting Heads:", probability_heads)
```

```python
import random

count_six = 0

for i in range(100):
    dice = random.randint(1, 6)
    
    if dice == 6:
        count_six += 1

experimental_probability = count_six / 100

print("Number of times 6 appeared:", count_six)
print("Experimental probability of getting 6:", experimental_probability)
```

```python
genres = ['pop', 'rock', 'pop', 'hiphop', 'pop',
          'rock', 'jazz', 'pop', 'hiphop', 'pop',
          'rock', 'pop', 'jazz', 'pop', 'rock',
          'pop', 'hiphop', 'rock', 'pop', 'jazz']

pop_count = genres.count('pop')
total_songs = len(genres)

probability = pop_count / total_songs

print("Number of pop songs:", pop_count)
print("Total songs:", total_songs)
print("Probability of picking a pop song:", probability)
```

```python
def both_ordered_probability(food_users, dessert_users):
    food_users = set(food_users)
    dessert_users = set(dessert_users)

    intersection = food_users.intersection(dessert_users)
    all_users = food_users.union(dessert_users)

    probability = len(intersection) / len(all_users)

    return probability


food_users = [1, 2, 3, 4, 5, 6]
dessert_users = [3, 4, 6, 7, 8]

probability = both_ordered_probability(food_users, dessert_users)

print("Probability that a user ordered both food and dessert:", probability)
```

```python
def conditional_probability(phone_users, headphone_users):
    phone_users = set(phone_users)
    headphone_users = set(headphone_users)

    both = phone_users.intersection(headphone_users)

    probability = len(both) / len(phone_users)

    return probability


phone_users = [1, 2, 3, 4, 5, 6]
headphone_users = [2, 3, 5, 7, 8]

probability = conditional_probability(phone_users, headphone_users)

print("Probability of buying headphones given phone purchase:", probability)
```
