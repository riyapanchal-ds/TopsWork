Que-1) Write down the null hypothesis (H0) and alternative hypothesis (H1) for the following scenario: You believe that people spend more time on Instagram than on YouTube daily.
Ans) Null Hypothesis (H0): People spend the same amount of time on Instagram and YouTube daily.
H0: μInstagram ≤ μYouTube
Alternative Hypothesis (H1): People spend more time on Instagram than on YouTube daily.
H1: μInstagram > μYouTube

```python
from scipy.stats import ttest_ind_from_stats
mean_A = 350
std_A = 50
n_A = 30
mean_B = 320
std_B = 50
n_B = 30

t_statistic, p_value = ttest_ind_from_stats(
    mean1=mean_A, std1=std_A, nobs1=n_A,
    mean2=mean_B, std2=std_B, nobs2=n_B,
    equal_var=True
)

print("t-statistic:", t_statistic)
print("p-value:", p_value)
```

# A/B Testing Example – Instagram
I will take Instagram's Reels button as an example.
## Control Group
In the control group (Group A), users will see the existing Reels button.
## Variant Group
In the variant group (Group B), users will see the new redesigned Reels button.
## Metric
I will measure the average number of Reels watched by each user in a day.
This will help me check whether the new Reels button increases user engagement.
## Null Hypothesis (H0)
There is no significant difference in the average number of Reels watched by users in Group A and Group B.
H0: μA = μB
## Alternative Hypothesis (H1)
There is a significant difference in the average number of Reels watched by users in Group A and Group B.
H1: μA ≠ μB
## A/B Test
I will randomly divide Instagram users into two groups.
Group A → Existing Reels button
Group B → New Reels button
Then I will collect the data and compare the average number of Reels watched by both groups.
## Decision
If p-value < 0.05, I will reject the null hypothesis.
If p-value >= 0.05, I will fail to reject the null hypothesis.
## Conclusion
If the p-value is less than 0.05, I can say that the new Reels button has a significant effect on the number of Reels watched by users.

# A/B Testing – Flipkart Wishlist Button
Suppose Flipkart tests a new "Add to Wishlist" button.
Out of 1000 users, 520 users clicked the new button and 480 users clicked the old button.
The difference is 40 users, or 4 percentage points. Without doing detailed calculations, this difference does not look very large, so it could be due to chance.
To decide whether the difference is statistically significant, we can use the p-value.
If the p-value is less than 0.05, we reject the null hypothesis and say the difference is statistically significant.
If the p-value is greater than or equal to 0.05, we fail to reject the null hypothesis and say the difference is likely due to chance.
Therefore, we need to calculate the p-value before making a final decision.
